```python
from nrem_analysis.constant import MOUSE_IDS_TTX, PROCESSED_DIR, RAW_DIR

import numpy as np
import pynapple as nap

from scipy.io import loadmat

for mouse_id in MOUSE_IDS_TTX:
    print(f"Processing: {mouse_id}")

    virtual_hd = []
    rem_decoded = loadmat(RAW_DIR / "ttx" / mouse_id / f"YutaTest{mouse_id}-eachREMdecode_1ms_results_HomeCage.mat")['eachREMdecode1ms_results']
    sleep = nap.load_file(PROCESSED_DIR / "ttx" / mouse_id / "sleep.npz")
    
    for idx in range(len(rem_decoded)):
        rem_decoded_ep = rem_decoded[idx]
        d, t = rem_decoded_ep[0].item()
        if (d.size > 0) and (t.size > 0):
            virtual_hd.append(nap.Tsd(t=t.squeeze(), d=d.squeeze()))
    
    virtual_hd = np.concatenate(virtual_hd)
    assert len(virtual_hd.time_support) == len(sleep[sleep['state'] == 'rem'])

    virtual_hd.save(PROCESSED_DIR / "ttx" / mouse_id / "rem_hd")
```
